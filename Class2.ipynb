 # DATA TYPES

## Number

## integer

```python
age = 20
print(f"My age is {age}")
```

## float

```python
CGPA = 4.0
print(f"My CGPA is {CGPA}")
```

## complex number

```python
x1 = 2 + 4j
x2 = 5 - 2j

print("First complex number ", x1)
print("Second complex number ", x2, "\n")

print("Operations")
print("Sum of x1 and x2", x1 + x2)
print("Subtraction of x1 and x2", x1 - x2)
```

# String

```python
name = "Awolaja olasubomi"
university = "Babcock University"

print(f"My name is {name}, I attend {university}.")
```

# Boolean

```python
is_male = True
print("Are you a male?", is_male)

x = 5
y =10
print("Is X greater than Y? ", x > y)
print("Is X less than Y? ", x < y)
print("Is X equal to Y? ", x == y)
```

# DATA STRUCTURES

# List

```python
names = ["Bola", "Subomi", "Femi"]
print(type(names))
print(names)
```

# Tuples

```python
age = (12, 23, 34, 45)
print(type(age))
print(age)
```

# Set

```python
gpa = {3.5, 4.5, 1.9}
print(type(gpa))
print(gpa)
```

## Dictionary

```python
student = {
    "name" : "Subomi",
    "age" : "20" ,
    "course" : "Information technology"
}


print(f"Name: {student["name"]}")
print(f"Age :{student["age"]}")
print(f"Course: {student["course"]}")
```
